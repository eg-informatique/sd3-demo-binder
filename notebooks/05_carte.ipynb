# 5. Carte simple avec Observable Plot

Plotly nécessite un token Mapbox pour les cartes choroplèthes. Pour une carte simple sans configuration supplémentaire, on utilise ici **Observable Plot** avec des données géographiques au format TopoJSON.

## 5.1 Charger les données géographiques

```typescript
import * as topojson from "topojson-client";
import * as Plot from "@observablehq/plot";
import { parseHTML } from "linkedom";

// Deno n'a pas de page web : linkedom fournit un « document » virtuel
// dans lequel Observable Plot peut dessiner.
const { document } = parseHTML("<!doctype html><html><body></body></html>");

const worldUrl =
  "https://cdn.jsdelivr.net/npm/world-atlas@2/countries-110m.json";
const world = await fetch(worldUrl).then((r) => r.json());
const countries = topojson.feature(world, world.objects.countries);
```

## 5.2 Afficher une carte du monde

```typescript
const svg = Plot.plot({
  document,
  projection: "equal-earth",
  marks: [
    Plot.geo(countries, { fill: "#e0e0e0", stroke: "#ffffff" }),
  ],
}).outerHTML;

await Deno.jupyter.display({ "text/html": svg }, { raw: true });
```

## 5.3 Ajouter des points pour quelques capitales

```typescript
const cities = [
  { name: "Paris", lon: 2.35, lat: 48.85 },
  { name: "Berlin", lon: 13.40, lat: 52.52 },
  { name: "Rome", lon: 12.50, lat: 41.90 },
  { name: "Madrid", lon: -3.70, lat: 40.42 },
];

const svgWithCities = Plot.plot({
  document,
  projection: "equal-earth",
  marks: [
    Plot.geo(countries, { fill: "#e0e0e0", stroke: "#ffffff" }),
    Plot.dot(cities, { x: "lon", y: "lat", r: 5, fill: "#ff6384" }),
    Plot.text(cities, {
      x: "lon",
      y: "lat",
      text: "name",
      dy: -10,
      fontSize: 10,
    }),
  ],
}).outerHTML;

await Deno.jupyter.display({ "text/html": svgWithCities }, { raw: true });
```
