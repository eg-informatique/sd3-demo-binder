# 1. Affichage et gestion de fichiers avec Deno

Ce notebook montre comment :
- afficher du texte, du markdown et du HTML ;
- télécharger un fichier CSV depuis Internet ;
- lire un fichier local avec Deno.

## 1.1 Afficher du texte

```typescript
// La dernière expression d'une cellule Deno est affichée automatiquement.
"Hello World"
```

## 1.2 Afficher du markdown

```typescript
await Deno.jupyter.display(
  { "text/markdown": "## Titre en markdown\n\nUn [lien](https://github.com)" },
  { raw: true },
);
```

## 1.3 Afficher du HTML (avec un bouton interactif)

```typescript
const buttonHtml = `
  <button id=\"monBouton\">Clique-moi</button>
  <script type=\"module\">
    const btn = document.getElementById('monBouton');
    btn.addEventListener('click', () => {
      btn.textContent = 'Tu as cliqué !';
    });
  </script>
`;

await Deno.jupyter.display({ "text/html": buttonHtml }, { raw: true });
```

## 1.4 Afficher une image

```typescript
const logoUrl = "https://nodejs.org/static/images/logo.svg";

await Deno.jupyter.display(
  { "text/html": `<img src="${logoUrl}" width="150" />` },
  { raw: true },
);
```

## 1.5 Télécharger un fichier CSV

On utilise l'API `fetch` de Deno, qui est disponible nativement.

```typescript
import { ensureDir } from "https://deno.land/std@0.224.0/fs/mod.ts";
import { dirname } from "https://deno.land/std@0.224.0/path/mod.ts";

const baseUrl = "https://raw.githubusercontent.com/datasciencedojo/datasets";
const csvUrl = `${baseUrl}/master/titanic.csv`;
const outPath = "../data/titanic.csv";

const response = await fetch(csvUrl);
if (!response.ok) {
  throw new Error(`HTTP ${response.status}`);
}

const csvText = await response.text();
await ensureDir(dirname(outPath));
await Deno.writeTextFile(outPath, csvText);

console.log(`Fichier téléchargé : ${outPath}`);
console.log(`Taille : ${csvText.length} caractères`);
```

## 1.6 Lire un fichier CSV localement

```typescript
const contenu = await Deno.readTextFile("../data/titanic.csv");
const lignes = contenu.split("\n").slice(0, 6);

for (const ligne of lignes) {
  console.log(ligne);
}
```
